```python
import warp as wp
wp.init()

print(wp.__version__)

@wp.kernel
def my_kernel(input: wp.array(dtype=float), dim: int, out : wp.array(dtype=float)):
    tid = wp.tid()
    factor = wp.float32(1.0/2.0) ** wp.float32(dim)
    out[tid] = input[tid] * factor
    
x = wp.array([1.0, 2.0, 3.0, 4.0], dtype=float)
y = wp.zeros_like(x)

wp.launch(kernel=my_kernel, dim=x.shape[0], inputs=[x, 4], outputs=[y])

print(y.numpy())
    
```
